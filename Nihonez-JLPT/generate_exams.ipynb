# Generate JLPT exams

Writes a markdown exam for each saved Nihonez HTML page in `.data/` that does not already have its markdown file under `outputs/<exam>/`. The file is named from the exam directory with hyphens between words, for example `JLPT-N4-Mock-Test-December-2024.md`. Referenced images are copied into `media/`. Audio files are left out; listening questions keep their transcripts.

```python
from nihonez_jlpt.pipeline import generate_missing

result = generate_missing()

print("Wrote:")
for name in result.written:
    print(f"  {name}")

print("Skipped:")
for name in result.skipped:
    print(f"  {name}")
```
